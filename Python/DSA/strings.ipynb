Given two strings s and p, return an array of all the start indices of p's anagrams in s. You may return the answer in any order.

Input: s = "cbaebabacd", p = "abc"
Output: [0,6]

```python
s = "cbaebabacd" ; p = "abc"
def anagram(s, p):
    indices = []
    p = sorted(p)
    for i in range(len(s)):
        s1 = s[i:i+len(p)]
        if p == sorted(s1):
            indices.append(i)
    return indices
anagram(s,p)
```

Given an array of strings strs, group the anagrams together. You can return the answer in any order.

Input: strs = ["eat","tea","tan","ate","nat","bat"]

Output: [["bat"],["nat","tan"],["ate","eat","tea"]]

```python
from collections import defaultdict
strs = ["eat","tea","tan","ate","nat","bat"]
def anagram(st):
    anagrm = defaultdict(list)
    for s in st:
        sort_s = ''.join(sorted(s))
        anagrm[sort_s].append(s)
    return list(anagrm.values())
anagram(strs)
```

Given a string s, sort it in decreasing order based on the frequency of the characters. The frequency of a character is the number of times it appears in the string.

Input: s = "tree"

Output: "eert"

```python
s1 = "tree"
def freq_count(st1):
    freq = {}
    for s1 in st1:
        if s1 not in freq:
            freq[s1] = 1
        else:
            freq[s1] += 1
    ans=""
    for k,v in sorted(freq.items(), key=lambda item: item[1]):
        for i in range(v):
            ans+=k
    return ans[::-1]

freq_count(s1)
```

```python
s3 = "lee(t(c)o)de)"

def minRemoveToMakeValid(s):

    stack = []
    new_str = list(s)
    to_close_cnt = 0
    for i in range(len(new_str)):
        if new_str[i] == '(' :
            stack.append(i)
            to_close_cnt+=1

        elif new_str[i] == ')':
            if stack and to_close_cnt > 0:
                stack.pop()
                to_close_cnt-=1
            else:
                stack.append(i)
        else:
            continue
    if stack:
        for i in  range(len(stack)-1, -1, -1):
            new_str[stack[i]]  = ''
    else:
        return s

    return ''.join(new_str)
minRemoveToMakeValid(s3)
```

```python
def remove_adc_dup(st):
    i = 0
    st = list(st)
    while i < len(st)-1:
        if st[i] == st[i+1]:
            st.pop(i)
            st.pop(i)
            i = 0
        else:
            i = i + 1
    return "".join(st)
remove_adc_dup('abbaca')
```

```python
#1. Program to reverse a given string without any string function.

def reverse_str(st):
    st = list(st)
    l = 0
    r = len(st)-1
    while l < r:
        st[l], st[r] = st[r], st[l]
        l = l+1
        r = r-1
    return "".join(st)
print(reverse_str("nabeel"))
```

```python
#2. Program to check if a given string is Palindrome or not.

def is_palindrom(st):
    st = st.lower()
    l  = 0
    r = len(st)-1
    while l<r:
        if st[l] == st[r]:
            l = l+1
            r = r-1
        else:
            return False
    return True

print(is_palindrom("Madam"))
```

```python
#3 names = {"Shashi","Ravi","Suman","Shashi","Suman"};

def cnt_freq(arr):
    output = {}
    for i in arr:
        if i not in output:
            output[i] = 1
        else:
            output[i] = output[i] + 1
    return output
print(cnt_freq(["Shashi","Ravi","Suman","Shashi","Suman"]))
```

```python
# 4. Program to reverse a given sentence

def rev_sent(sent):
    words = sent.split()
    l = 0
    r = len(words)-1
    while l < r:
        words[l], words[r] = words[r], words[l]
        l = l+1
        r = r-1
    return " ".join(words)
print(rev_sent("My Name Is Shashi"))
```

```python
# Number to word conversion

def no_to_word(n):
    n = str(n)
    num_map = {"1": "one", "2":"two", "3":"three", "4":"four", "5":"five",
              "6":"six", "7":"seven", "8":"eight", "9":"nine", "0":"zero"}
    output = ""
    for i in n:
        if i in num_map:
            output = output + ' ' + num_map[i]
    return output
print(no_to_word("007"))
print(no_to_word(1234))
```

```python
def simple_number_to_words(num):

    ones = ["Zero", "One", "Two", "Three", "Four", "Five",
            "Six", "Seven", "Eight", "Nine"]
    teens = ["Ten", "Eleven", "Twelve", "Thirteen", "Fourteen",
             "Fifteen", "Sixteen", "Seventeen", "Eighteen", "Nineteen"]
    tens = ["", "", "Twenty", "Thirty", "Forty", "Fifty",
            "Sixty", "Seventy", "Eighty", "Ninety"]

    if num < 10:
        return ones[num]

    if 10 <= num < 20:
        return teens[num - 10]

    # to handle 21-100
    if 20 <= num < 100:
        tens_word = tens[num // 10]
        ones_word = ones[num % 10]

        if ones_word == "Zero":
            return tens_word
        return tens_word + " " + ones_word

    # to handle 100-999
    if 100 <= num < 1000:
        # to get hundreds digit
        hundreds_digit = num // 100
        hundreds_word = ones[hundreds_digit] + " Hundred"

        # to get remainder (last two digits)
        remainder = num % 100

        if remainder == 0:
            return hundreds_word

        remainder_words = simple_number_to_words(remainder)
        return hundreds_word + " " + remainder_words

    # to handle 1000-9999
    if 1000 <= num < 10000:
        thousands_digit = num // 1000
        thousands_word = ones[thousands_digit] + " Thousand"

        remainder = num % 1000

        if remainder == 0:
            return thousands_word

        remainder_words = simple_number_to_words(remainder)
        return thousands_word + " " + remainder_words



print(simple_number_to_words(7))
print(simple_number_to_words(1234))
```

```python
def largest_prime(num):
    fact = 2
    while num > 1:
        if num % fact == 0:
            num = num //fact
        else:
            fact = fact + 1
    return fact
largest_prime(56)
```
